# Test problem 1: Cooling of a rod (cooling benchmark)

A rod of length $L$ cools with rate $k$ from temperature 1. We integrate
with explicit Euler steps of size $dt$ until $T_{end}$ and compare the
final temperature with $e^{-k T_{end}}$.

```python
import math
%matplotlib inline
```

```python
L = 1.0  # length of the rod (m)
k = 0.5  # cooling rate (1/s)
T_end = 2.0  # end time (s)
dt = 1 / 100  # time step (s)
origin = (0.0, 0.0)
rate = float(max(0.25, 0))
num_steps = int(T_end / dt)
```

```python
temperature = []
times = []
u = 1.0
for i in range(num_steps):
    u = u - dt * k * u
    temperature.append(u)
    times.append((i + 1) * dt)
error = abs(u - math.exp(-k * T_end))
print(f"Final temperature: {u}")
print("error", error)
```
